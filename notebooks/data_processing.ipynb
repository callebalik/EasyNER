## Setting entity ERROR labels - AMBIG, MISLAB, MISSP

```python
import pandas as pd

# Get error label CSV as dataframe and sort columns
relative_path = '../dictionaries/misslabeled_ner.csv'
error_labels_df = pd.read_csv(relative_path)

# Clean whitespace from error_id column
error_labels_df['error_id'] = error_labels_df['error_id'].str.strip()
error_labels_df['entity_text'] = error_labels_df['entity_text'].str.strip()
error_labels_df['entity_type'] = error_labels_df['entity_type'].str.replace(' ', '_')
# Sort by multiple columns including error_id, entity_type, and entity_text
error_labels_df_sorted = error_labels_df.sort_values(by=['error_id', 'entity_type', 'entity_text'])

# Reorder columns
error_labels_df_sorted = error_labels_df_sorted[['error_id', 'entity_type', 'entity_text']]

# Change column names
error_labels_df_sorted.columns = ['Error Code', 'Named Entity Class', 'Entity Text']

# Remove index column
error_labels_df_sorted.reset_index(drop=True, inplace=True)

error_labels_df_sorted
```
